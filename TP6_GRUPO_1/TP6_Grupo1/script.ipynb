# [dbo].[FiltraProductos]

```sql
/****** Object:  StoredProcedure [dbo].[FiltraProductos]    Script Date: 20/05/2024 20:24:48 ******/
SET ANSI_NULLS ON
GO
SET QUOTED_IDENTIFIER ON
GO
CREATE PROCEDURE [dbo].[FiltraProductos]
    @IdCategoria int,
    @IdProducto int
AS
BEGIN
    SELECT * FROM Productos
    WHERE (@IdCategoria IS NULL OR @IdCategoria = @IdCategoria)
    AND (@IdProducto IS NULL OR IdProducto = @IdProducto)
END
GO
```

# [dbo].[FiltrarProductos]

```sql
/****** Object:  StoredProcedure [dbo].[FiltrarProductos]    Script Date: 20/05/2024 20:24:48 ******/
SET ANSI_NULLS ON
GO
SET QUOTED_IDENTIFIER ON
GO
CREATE PROCEDURE [dbo].[FiltrarProductos]
    @IdCategorÍa int,
    @IdProducto int
AS
BEGIN
    SELECT * FROM Productos
    WHERE (@IdCategorÍa IS NULL OR IdCategorÍa = @IdCategorÍa)
    AND (@IdProducto IS NULL OR IdProducto = @IdProducto)
END
GO
```

# [dbo].[FiltrarPructos]

```sql
/****** Object:  StoredProcedure [dbo].[FiltrarPructos]    Script Date: 20/05/2024 20:24:48 ******/
SET ANSI_NULLS ON
GO
SET QUOTED_IDENTIFIER ON
GO
Create Procedure [dbo].[FiltrarPructos]
(@IdProducto int, @IdCategoría int)
AS
BEGIN
    SELECT * FROM Productos
    WHERE (@IdProducto IS NULL OR IdProducto = @IdProducto)
    AND (@IdCategoría IS NULL OR IdCategoría = @IdCategoría) 
END
GO
```

# [dbo].[spActualizarProducto]

```sql
/****** Object:  StoredProcedure [dbo].[spActualizarProducto]    Script Date: 20/05/2024 20:24:48 ******/
SET ANSI_NULLS ON
GO
SET QUOTED_IDENTIFIER ON
GO

CREATE PROCEDURE [dbo].[spActualizarProducto]
(
@IDPRODUCTO INT, 
@NOMBREPRODUCTO VARCHAR(40),
@CANTPRODXUNI VARCHAR(20),
@PRECIOXUNIDAD MONEY
)
AS 
UPDATE Productos
SET
NombreProducto = @NOMBREPRODUCTO, 
CantidadPorUnidad = @CANTPRODXUNI, 
PrecioUnidad = @PRECIOXUNIDAD
WHERE IdProducto = @IDPRODUCTO

RETURN
GO
```
